Using Flickr, we can get all the court designs from 2020 back to the beginning of time. 
Steps:
    * add them to personal gallery based on unique designs
    * load them here
    * upload to roboflow

```python
import os
import re
import requests
import flickrapi
from dotenv import load_dotenv
from tqdm import tqdm  # For progress bar

# Print the current working directory (optional for debugging)
print(os.getcwd())

# Set the path to your .env file relative to the current working directory
dotenv_path = os.path.join(os.getcwd(), '../.env')
load_dotenv(dotenv_path)

# Get Flickr API credentials from environment variables
api_key = os.getenv('FLICKR_API_KEY')
api_secret = os.getenv('FLICKR_API_SECRET')

# Initialize Flickr API client
flickr = flickrapi.FlickrAPI(api_key, api_secret, format='parsed-json')

# Directory to save images
output_dir = '../../data/images/flickr_images'
os.makedirs(output_dir, exist_ok=True)

# Function to sanitize file names
def sanitize_filename(name):
    return re.sub(r'[\\/*?:"<>|]', '_', name)

# Example: Fetch photos from a gallery and download them
def get_gallery_photos(gallery_id, per_page=10):
    try:
        photos = flickr.galleries.getPhotos(gallery_id=gallery_id, per_page=per_page)
        photo_list = photos['photos']['photo']
        for photo in tqdm(photo_list, desc="Downloading photos"):
            title = photo.get('title', 'untitled').replace(' ', '_')
            title = sanitize_filename(title)  # Sanitize the title for safe filenames
            photo_id = photo.get('id')
            farm = photo.get('farm')
            server = photo.get('server')
            secret = photo.get('secret')
            # Construct the URL for each photo
            url = f'https://farm{farm}.staticflickr.com/{server}/{photo_id}_{secret}.jpg'

            # Download the image
            response = requests.get(url, stream=True)
            if response.status_code == 200:
                # Save the image
                file_path = os.path.join(output_dir, f"{title}_{photo_id}.jpg")
                with open(file_path, 'wb') as file:
                    for chunk in response.iter_content(1024):
                        file.write(chunk)
                print(f"Downloaded: {file_path}")
            else:
                print(f"Failed to download: {url} (Status code: {response.status_code})")
    except Exception as e:
        print(f"Error fetching gallery photos: {e}")

if __name__ == "__main__":
    # Replace with your own gallery ID
    gallery_id = '72157723176679332'  # Example gallery ID
    get_gallery_photos(gallery_id)
```
