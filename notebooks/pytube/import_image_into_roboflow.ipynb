```python
import yt_dlp

def download_video_yt_dlp(url, output_path="downloads"):
    ydl_opts = {"outtmpl": f"{output_path}/%(title)s.%(ext)s"}
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([url])

# Example usage
download_video_yt_dlp("https://www.youtube.com/watch?v=c7n8G4wK2Ss")


# Example usage
output_path="../../data/images/roboflow_input/pytube_images"
url = "https://www.youtube.com/watch?v=fhOOssqKLas"
download_video_yt_dlp(url, output_path=output_path)
```

```python
import cv2
import os

def extract_frames(video_path, output_folder="frames", frame_rate=1):
    os.makedirs(output_folder, exist_ok=True)
    cap = cv2.VideoCapture(video_path)
    count = 0
    frame_count = 0

    # Set the desired frame rate (e.g., 1 frame every 1 second)
    fps = cap.get(cv2.CAP_PROP_FPS)
    frame_interval = int(fps / frame_rate)

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        if count % frame_interval == 0:
            frame_path = os.path.join(output_folder, f"frame_{frame_count:04d}.jpg")
            cv2.imwrite(frame_path, frame)
            frame_count += 1
        count += 1

    cap.release()
    print(f"Extracted {frame_count} frames to {output_folder}")
video_path = "../../data/images/roboflow_input/pytube_images/Stephen Curry threes but they get increasingly more ridiculous.mp4"
# Example usage
extract_frames(video_path=video_path, output_folder="../../data/images/roboflow_input/pytube_images", frame_rate=1)  # 1 frame per second
```
